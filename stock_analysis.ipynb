```python
import yfinance as yf
import pandas as pd

stock = yf.download("RELIANCE.NS", start="2023-01-01")

stock.head()
```

```python
import matplotlib.pyplot as plt

stock["Close"].plot(figsize=(12,5))

plt.title("Reliance Stock Price Trend")
plt.xlabel("Date")
plt.ylabel("Price")

plt.show()
```

```python
stock["Daily Return"] = stock["Close"].pct_change()*100

stock[["Close","Daily Return"]].head()
```

```python
best_day = stock["Daily Return"].max()
worst_day = stock["Daily Return"].min()

print("Best Day Return:", best_day)
print("Worst Day Return:", worst_day)
```

```python
stock["MA50"] = stock["Close"].rolling(50).mean()

stock[["Close","MA50"]].tail()
```

```python
plt.figure(figsize=(12,5))

plt.plot(stock["Close"], label="Close Price")
plt.plot(stock["MA50"], label="50 Day MA")

plt.title("Reliance Stock vs Moving Average")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.plot(stock["Close"], label="Close Price")
plt.plot(stock["MA50"], label="50 Day MA")

plt.title("Reliance Stock vs Moving Average")
plt.legend()

plt.savefig("reliance_stock_analysis.png")

plt.show()
```
